```python
from enum import IntEnum
import numpy as np

# 1. THE ACTION SPACE (Discrete)
# We limit the agent to 3 simple actions to ensure stability.
# 0 = Do Nothing
# 1 = Buy 1 Unit (Aggressively)
# 2 = Sell 1 Unit (Aggressively)

class Actions(IntEnum):
    HOLD = 0
    BUY = 1
    SELL = 2

# 2. THE STATE SPACE (Observation)
# The "State" is what the agent sees. It must be normalized.
# We will use a vector of size 14 for this project.

# State Vector Structure:
# [0-4]   : Top 5 Bid Prices (Normalized relative to Mid-Price)
# [5-9]   : Top 5 Ask Prices (Normalized relative to Mid-Price)
# [10]    : Portfolio Inventory (Normalized)
# [11]    : Portfolio Cash / PnL (Normalized)
# [12]    : Bid-Ask Spread (Normalized)
# [13]    : Market Mid-Price Volatility (Rolling)

OBSERVATION_DIMS = 14

# Normalization Constants (To keep Neural Network inputs between -1 and 1)
MAX_PRICE_DEVIATION = 0.05   # We assume price won't deviate >5% from reference in short term
MAX_INVENTORY_LIMIT = 100    # Max positions agent can hold
MAX_CASH_LIMIT = 1_000_000   # Normalization factor for cash

# 3. REWARD HYPERPARAMETERS
# The "Reward" tells the agent what we value.
# Formula: Reward = (Incremental PnL) - (Risk Penalty)

# How much we penalize holding inventory (Inventory Risk)
# Higher = Agent closes positions faster. Lower = Agent holds longer.
INVENTORY_RISK_COEFF = 0.01

# How much we penalize losing money (Drawdown Risk)
# If the agent loses money, we punish it 1.5x more than we reward making money.
# This teaches "Capital Preservation".
DOWNSIDE_PENALTY_MULT = 1.5

# Transaction Cost (Per trade)
# To prevent "churning" (buying/selling uselessly).
TRANSACTION_COST = 0.50 

# ==========================================
# 4. ENVIRONMENT SETTINGS
# ==========================================

INITIAL_CASH = 100_000.0
MAX_STEPS_PER_EPISODE = 2000  # How long one "game" lasts
RANDOM_SEED = 42

def print_mdp_summary():
    print("--- Week 3 Day 1: MDP Configuration ---")
    print(f"Action Space: {len(Actions)} Discrete Actions ({[a.name for a in Actions]})")
    print(f"State Space:  {OBSERVATION_DIMS} Dimensions (Prices, Inventory, Cash, Volatility)")
    print(f"Reward Logic: PnL - ({INVENTORY_RISK_COEFF} * Inventory) - Transaction Costs")
    print("---------------------------------------")

if __name__ == "__main__":
    print_mdp_summary()
```
